```python
def nilai(huruf=""):
    data = {
        "A": 4.00,
        "A-": 3.75,
        "B+": 3.50,
        "B": 3.00,
        "B-": 2.75,
        "C+": 2.50,
        "C": 2.00,
        "C-": 1.75,
        "D": 1.50,
        "E": 1.25}
    return data.get(huruf, None)
total = 0
jumlah = 0

while True:
    huruf = input("masukkan nilai: ").upper()

    if huruf == "":
        break

    angka = nilai(huruf)

    if angka is not None:
        print("nilai =", angka)
        total += angka
        jumlah += 1
    else:
        print("Nilai tidak valid")

if jumlah > 0:
    print("Rata-ratanya adalah:", total / jumlah)
```

```python
def tahun_kabisat(tahun):
    if tahun % 400 == 0:
        return True
    elif tahun % 100 == 0:
        return False
    elif tahun % 4 == 0:
        return True
    else:
        return False

def jumlah_hari(bulan, tahun):
    if bulan == 2:
        if tahun_kabisat(tahun):
            return 29
        else:
            return 28
    elif bulan in [4, 6, 9, 11]:
        return 30
    else:
        return 31

print("Program ini akan menentukan jumlah hari dalam bulan tertentu")

while True:
    print("Masukkan 0 untuk menghentikan program")
    bulan = int(input("masukkan bulan (1-12): "))

    if bulan == 0:
        print("Terima kasih sudah menggunakan program saya, sampai berjumpa lagi.")
        break

    if bulan < 1 or bulan > 12:
        print("Bulan tidak valid, masukkan angka 1 sampai 12.")
        continue

    tahun = int(input("masukkan tahun (e.g., 2021): "))
    hari = jumlah_hari(bulan, tahun)
    print(hari, "hari dalam sebulan")
```

```python
def cek_prima(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

angka = int(input("Masukkan angka: "))
if cek_prima(angka):
    print(f"{angka} adalah bilangan Prima")
else:
    print(f"{angka} bukanlah bilangan Prima")
```

```python
def tentukan_ordinal(n):
    if 11 <= n % 100 <= 13:
        suffix = 'th'
    elif n % 10 == 1:
        suffix = 'st'
    elif n % 10 == 2:
        suffix = 'nd'
    elif n % 10 == 3:
        suffix = 'rd'
    else:
        suffix = 'th'

    return (n, suffix)

print("Ordinal Number (ketik 0 untuk berhenti)")
while True:
    angka = int(input("masukkan angka: "))
    print(tentukan_ordinal(angka))

    if angka == 0:
        print("terima kasih telah menggunakan program saya")
        break
```
